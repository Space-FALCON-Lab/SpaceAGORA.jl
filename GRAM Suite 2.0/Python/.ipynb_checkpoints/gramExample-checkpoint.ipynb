# Using the NASA GRAM Suite models with Python

## Preliminaries

This tutorial assumes that you have followed the instructions in the README to build the GRAM Suite and the  Python interface on your computer. 

As a general Python best practice, it is recommended to use a virtual environment to help manage dependencies. 

## Simple example:

First, import the gram module:

```python
from GRAMpy import gram
```

Next, we are going to get some data for a point in the Mars atmosphere on a specific day. This is similar to the C++ example shown in `Mars/examples/Mars.cpp`, if you would like to compare with the C++ interface. 

```python
inputParameters = gram.MarsInputParameters()

# Set the path to the Mars data folder- this is important! 
inputParameters.dataPath = "../Mars/data/"

# Load SPICE data
gram.NamelistReader().tryGetSpicePath(inputParameters)
gram.SpiceLoader().setInputParameters(inputParameters)

# Create Mars atmosphere 
mars = gram.MarsAtmosphere()
mars.setInputParameters(inputParameters)

# Set model options
mars.setMapYear(0)
mars.setMGCMDustLevels(2.0, 0.0, 0.0)
mars.setPerturbationScales(1.5)
mars.setMinRelativeStepSize(0.5)
mars.setSeed(1001)

# Set simulation start time 
time = gram.GramTime()
time.setStartTime(2020, 11, 2, 0, 0, 0, gram.UTC, gram.ERT)
mars.setStartTime(time)

# Set simulation position 
position = gram.Position()
position.height = 2
position.latitude = 22
position.longitude = 48 
position.elapsedTime = 100
position.isPlanetoCentric = True 
mars.setPosition(position)

# Update the model based on the current time and position 
mars.update()

# Output some information about the Mars atmosphere here
print("\nCurrent position: Lat " + str(mars.getPosition().latitude) + " E, Long " \
      + str(mars.getPosition().longitude) + " N. Height: " + str(mars.getPosition().height) + "km")
atmos = mars.getAtmosphereState()
print("Temperature: " + str(atmos.temperature))
print("Density: " + str(atmos.density))
print("Carbon dioxide mole fraction: " + str(atmos.carbonDioxide.moleFraction))
print("EW Winds: " + str(atmos.ewWind))

# Output Mars specific information with the MarsAtmosphereState:
print("Dust optical depth: " + str(mars.getMarsAtmosphereState().dustOpticalDepth))

```

## Plotting example

Using the Python interface makes it easier to quickly visualize your data, and explore it interactively. Here, we will make a quick plot to show atmospheric density vs. height above the surface 

### Plotting setup

```python
%matplotlib inline
```

```python
import matplotlib.pyplot as plt
import matplotlib.cm, matplotlib.colors
import numpy as np
```

### Density vs height above surface

```python
# Using the same Mars and Position objects we initialized above:
densities = []

position.height = 0

while position.height < 100:
    mars.setPosition(position)
    mars.update()
    densities.append(mars.getAtmosphereState().density)
    position.height += 0.1


fig, ax = plt.subplots()
ax.plot(densities, np.arange(0, 100.1, 0.1))
ax.set_xlabel("Density (kg/m^3)")
ax.set_ylabel("Height above surface (km)")
plt.show()
```

Well, that is not particularly interesting: atmospheric density is highest near the surface, and decreases with altitude. Luckly, with the Python interface we can easily extend the code above to explore several different attributes. The code snippet below follows the same structure, but collects data for four different plots.

```python
pressures = []

temperatures = []

ewWinds = []
ewDeviation = []

argon = []
carbonDioxide = []
carbonMonoxide = []
dihydrogen = []
dinitrogen = []
dioxygen = []
helium = []
hydrogen = []
oxygen = []
water = []

position.height = 0

# Sample every 100 meters vertically, starting at ground level and going up to 600km
while position.height < 600:

    mars.setPosition(position)
    mars.update()
    atmos = mars.getAtmosphereState()

    pressures.append(atmos.pressure)
    
    temperatures.append(atmos.temperature)
    
    ewWinds.append(atmos.ewWind)
    ewDeviation.append(atmos.ewStandardDeviation)

    argon.append(atmos.argon.massFraction)
    carbonDioxide.append(atmos.carbonDioxide.massFraction)
    carbonMonoxide.append(atmos.carbonMonoxide.massFraction)
    dihydrogen.append(atmos.dihydrogen.massFraction)
    dinitrogen.append(atmos.dinitrogen.massFraction)
    dioxygen.append(atmos.dioxygen.massFraction)
    helium.append(atmos.helium.massFraction)
    hydrogen.append(atmos.hydrogen.massFraction)
    oxygen.append(atmos.oxygen.massFraction)
    water.append(atmos.water.massFraction)

    position.height += 0.1

# Plot setup
fig, axs = plt.subplots(nrows=2, ncols=2)
vertical_profile = np.arange(0, 600, 0.1)

# Pressure
axs[0,0].plot(pressures, vertical_profile)
axs[0,0].set_xlabel("Pressure (pa)")
axs[0,0].set_ylabel("Height above surface (km)")
# Use logarithmic scale
axs[0,0].set_xscale("log")

# Temperature
axs[0,1].plot(temperatures, vertical_profile)
axs[0,1].set_xlabel("Temperature (K)")
axs[0,1].set_ylabel("Height above surface (km)")

# EW winds with +/- 3 standard deviation range
ewWinds = np.array(ewWinds)
ewDeviation = np.array(ewDeviation)
ewWindsLow = ewWinds - (3 * ewDeviation)
ewWindsHigh = ewWinds + (3 * ewDeviation)

axs[1,0].plot(ewWinds, vertical_profile)
axs[1,0].set_xlabel("East-West winds (m/s)")
axs[1,0].set_ylabel("Height above surface (km)")
axs[1,0].fill_betweenx(vertical_profile, ewWindsLow, ewWindsHigh, alpha=0.2, label="+/- 3 std")
axs[1,0].set_ybound(0, 200)

# Constituent mass percentages
axs[1,1].plot(argon, vertical_profile, label = "Ar")
axs[1,1].plot(carbonDioxide, vertical_profile, label = "CO2")
axs[1,1].plot(carbonMonoxide, vertical_profile, label="CO")
axs[1,1].plot(dihydrogen, vertical_profile, label="H2")
axs[1,1].plot(dinitrogen, vertical_profile, label="N2")
axs[1,1].plot(dioxygen, vertical_profile, label="O2")
axs[1,1].plot(helium, vertical_profile, label="He")
axs[1,1].plot(hydrogen, vertical_profile, label="H")
axs[1,1].plot(oxygen, vertical_profile, label="O")
axs[1,1].plot(water, vertical_profile, label="H2O")
axs[1,1].set_ylabel("Height above surface (km)")
axs[1,1].set_xlabel("Constituent mass percentages (%)")

# Output formatting
axs[1,0].legend()
axs[1,1].legend(loc='upper right', fontsize='xx-small')
plt.tight_layout()
plt.show()
```

# Advanced examples

## 1. Visualizing the topography of the Tharsis quadrangle

The Tharsis quadrangle on Mars stretches from 225 to 270 degrees east longitude and 0 to 30 degrees north latitude. It includes Olympus Mons, Ascreaus Mons, and Pavonis Mons.

For our next few examples, we are going to look at the atmosphere over this area. First, though, let's take a look at the topography to confirm we are in the right place and get a sense for how ground features could effect some of our data. For ease of formatting, we are going to look at a square area that encompasses all of the rectangular Tharsis quadrangle as well as the area from 30 to 45 degrees north of that rectangle.

```python
# Define target area
START_LAT = 45
END_LAT = 0
START_LONG = 225
END_LONG = 270

# Set number of sampling points 
LAT_POINTS = abs(START_LAT - END_LAT) * 2
LONG_POINTS = abs(START_LONG - END_LONG) * 2

# Initialize empty numpy array
data = np.zeros((LAT_POINTS, LONG_POINTS), float)

# Enumerate sample coordinates- setting up an even distribution of 
# our desired number of points over our target area
lats = np.linspace(START_LAT, END_LAT, LAT_POINTS)
lons = np.linspace(START_LONG, END_LONG, LONG_POINTS)

# Use the MOLA topographic data
mars.setMOLAHeights(True)
position = gram.Position() 
position.height = 0

# Collect data
for i in range(LAT_POINTS):
    for j in range(LONG_POINTS):
        position.setLatitude(lats[i])
        position.setLongitude(lons[j])
        mars.setPosition(position)
        data[i][j] = mars.getPosition().surfaceHeight

# Plotting setup
fig, ax = plt.subplots(figsize=(10,10), subplot_kw={"projection": "3d"})
X, Y = np.meshgrid(lons, lats)
ax.plot_surface(X,Y,data, color=matplotlib.colors.cnames["darkorange"])
ax.set_zlim(-3, 400) # Note elevation is not to scale
ax.set_xlabel("Longitude (deg. East)")
ax.set_ylabel("Latitude (deg. North)")
ax.set_zlabel("Elevation (km)")
```

The figure above shows our topographical aproximation of the surface of mars in this area. This is a good sanity check before we move on to visualizing atmospheric data. From left to right, we see 3 peaks: Olympus Mons, Pavonis Mons, and Ascraeus Mons, all matching up with actual topographical maps of the Tharsis quadrangle (for example, https://www.usgs.gov/maps/topographic-map-tharsis-quadrangle-mars). Keep in mind that the lightweight MOLA topographic dataset included in MarsGRAM only has half-degree resolution, so we should not expect to be getting very precise topography. Additionally, the elevation is not shown to scale. 

## 2. Winds example

In this example, we will plot a vector field of wind directions 5km above the surface of the Tharsis quadrangle. 

```python
# Set sample points
LAT_POINTS, LONG_POINTS = 20, 20
lats = np.linspace(START_LAT, END_LAT, num=LAT_POINTS, dtype=float)
lons = np.linspace(START_LONG, END_LONG, num=LONG_POINTS, dtype=float)
X, Y = np.meshgrid(lons, lats)

#Initialize empty array
ewWinds = np.zeros((LAT_POINTS, LONG_POINTS), float)
nsWinds = np.zeros((LAT_POINTS, LONG_POINTS), float)

# Set position 5km above surface
position.height = 5
mars.setPosition(position)

for i in range(LAT_POINTS):
    for j in range(LONG_POINTS):
        position.setLatitude(lats[i])
        position.setLongitude(lons[j])
        mars.setPosition(position)
        mars.update()
        ewWinds[i][j] = mars.getAtmosphereState().ewWind
        nsWinds[i][j] = mars.getAtmosphereState().nsWind

fig, ax = plt.subplots()
ax.set_title("Winds 5km above surface")
ax.set_xlabel("Longitude (deg. East)")
ax.set_ylabel("Latitude (deg. North)")
ax.quiver(X, Y, ewWinds, nsWinds)
```

## 3. Tornado plot example

In this example, we will visualize perturbations in the model's east-west winds over a large number of runs. **This may take a while- please be patient**

```python
RUNS = 1000
MAX_ALTITUDE = 200
STEP = 1
VERTICAL = np.arange(0, MAX_ALTITUDE, STEP)

ewWinds = []
ewDeviations = []
perturbedWinds = []

fig, ax = plt.subplots()
ax.set_xlabel("EW Winds (m/s)")
ax.set_ylabel("Height above surface (km)")
ax.set_title(str(RUNS) + " East West Wind Monte Carlo Runs")

# Get mean and standard deviations for this position
position = gram.Position()
position.latitude = 30
position.longitude = 130 
position.elapsedTime = 100
position.height = 0

while position.height < MAX_ALTITUDE:
    mars.setPosition(position)
    mars.update()
    ewWinds.append(mars.getAtmosphereState().ewWind)
    ewDeviations.append(mars.getAtmosphereState().ewStandardDeviation)
    position.height += STEP

# Get perturbed winds
for i in range(RUNS):
    position.height = 0
    perturbedWinds = []
    while position.height < MAX_ALTITUDE:
        mars.setPosition(position)
        mars.update()
        perturbedWinds.append(mars.getAtmosphereState().perturbedEWWind)
        position.height += STEP
    
    ax.plot(perturbedWinds, VERTICAL, color="coral", linewidth=.25)

ewWinds = np.array(ewWinds)
ewDeviations = np.array(ewDeviations)
ax.plot(ewWinds, VERTICAL, label="Mean")
ax.fill_betweenx(VERTICAL, ewWinds - 3 * ewDeviations, ewWinds + 3 * ewDeviations, label="+/- 3 std", alpha = .2)
ax.legend()


```

## 4. Dust Storm GIF

In the next example, we will visualize the dust storm model that MarsGRAM uses to simulate dust events. Looking at the same area of mars we just mapped, we will make an animation showing the dust optical depth over the course of a storm. The code below will generate a gif showing the lifecycle of a dust storm. Unlike the other plots, this gif cannot be displayed inline and will instead be saved in this folder. 

**This code cell may take a few minutes to generate the animation.** Once it is done you will be able to open up `dust_storm.gif` to see the result.

```python
# Dust storm gif 
import matplotlib.animation as animation

TIME_STEPS = 50
STORM_DURATION = 24
STORM_LAT = 20
STORM_LON = 240
LAT_POINTS = abs(START_LAT - END_LAT) * 2
LONG_POINTS = abs(START_LONG - END_LONG) * 2

lats = np.linspace(START_LAT, END_LAT, LAT_POINTS)
lons = np.linspace(START_LONG, END_LONG, LONG_POINTS)

# Set  up plot
fig, ax = plt.subplots()
ax.set_xticks([])
ax.set_yticks([])
ax.set_title("Medium Duration Dust Storm Over Tharsis Quadrangle")

#Initialize empty array
dust = np.zeros((LAT_POINTS, LONG_POINTS), float)
frames = []

position = gram.Position()
position.height = 0
position.elapsedTime = 0
mars.setPosition(position)
mars.update()

mars.setDustStorm(mars.getEphemerisState().longitudeSun, STORM_DURATION, intensity=2.5, maxRadius=400, lat=STORM_LAT, lon=STORM_LON)

# Generate image frame at each time step
for t in range(TIME_STEPS):
    position.elapsedTime += (3.15e7 / 360) # aprox 1 degree Ls, converted to seconds
    mars.setPosition(position)
    mars.update()

    # Iterate over sample points at this time
    for i in range(LAT_POINTS):
        for j in range(LONG_POINTS):
            position.setLatitude(lats[i])
            position.setLongitude(lons[j])
            mars.setPosition(position)
            mars.update()
            dust[i][j] = mars.getMarsAtmosphereState().dustOpticalDepth

    # Draw frame
    im = ax.imshow(dust, animated=True, cmap=matplotlib.cm.Oranges)
    frames.append([im])   

# plot setup
plt.colorbar(frames[int(TIME_STEPS / 2)][0], label="Optical dust depth")
ani = animation.ArtistAnimation(fig, frames)
ani.save("dust_storm.gif")
```
